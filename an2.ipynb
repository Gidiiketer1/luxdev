```python
import pandas as pd
import scipy.stats as stats
import statsmodels.api as sm
from statsmodels.formula.api import ols
```

```python
house_prices = pd.read_csv(r"house_prices.csv")
house_prices.head()
```

```python
house_prices.info()
```

```python
house_prices['condition'].value_counts()
```

```python
house_prices['view'].value_counts()
```

```python
house_prices['waterfront'].value_counts()
```

```python
house_prices['grade'].value_counts()
```

```python
house_prices.columns
```

```python
formula = "price ~ bedrooms + bathrooms + sqft_living + sqft_lot + floors + waterfront + view + condition + grade + year_built + year_renovated + C(waterfront) + C(condition) + C(grade)"
lm = ols(formula=formula, data=house_prices).fit()
table = sm.stats.anova_lm(lm, typ=2)
print(table)
```
