```python
import pandas as pd
import scipy.stats as stats
import statsmodels.api as sm
from statsmodels.formula.api import ols
import numpy as np
```

```python
homepage_action  = pd.read_csv(r'homepage_actions.csv')
homepage_action.head()
```

```python
experimental_group = homepage_action[homepage_action['group'] == "experiment"]
experimental_group.head()
```

```python
click_action = experimental_group['action'].value_counts()
click_action
```

```python
control_group = homepage_action[homepage_action['group'] == "control"]
control_group.head()
```

```python
click_action_control = control_group['action'].value_counts()
click_action_control
```

```python
experimental_views = (experimental_group["action"] == "view").sum()
experimental_views
```

```python
experimental_clicks = (experimental_group["action"] == "click").sum()
experimental_clicks
```

```python
experimental_ctr = (experimental_clicks / experimental_views)
experimental_ctr
```

```python
control_views = (control_group["action"] == "view").sum()
control_views
```

```python
control_clicks = (control_group["action"] == "click").sum()
control_clicks
```

```python
control_ctr = (control_clicks / control_views)
control_ctr
```

```python
e_sum = experimental_group["action"].value_counts()
e_sum
```

```python
experimental_mean = (experimental_views*experimental_ctr)
experimental_mean
```

```python
control_mean = (control_views*control_ctr)
control_mean
```

```python
# compute two-sided p-value for difference in proportions (CTR)
p1 = experimental_ctr
p2 = control_ctr
n1 = experimental_views
n2 = control_views

if n1 == 0 or n2 == 0:
	raise ValueError("One of the groups has zero views; cannot compute p-value.")

# pooled proportion
p_pool = (experimental_clicks + control_clicks) / (n1 + n2)

# standard error for difference in proportions
se = np.sqrt(p_pool * (1 - p_pool) * (1/n1 + 1/n2))

# z-statistic
z = (p1 - p2) / se

# two-sided p-value
p_value = 2 * (1 - stats.norm.cdf(abs(z)))
p_value
```
